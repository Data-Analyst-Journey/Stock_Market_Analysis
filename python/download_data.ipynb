```python
# ============================================
# Stock Data Downloader
# Author: Samaneh Kavianfar
# Project: Stock Market Analysis
# ============================================

import yfinance as yf
import pandas as pd
from datetime import datetime, timedelta

# ============================================
# 1. Define the list of companies
# ============================================
companies = {
    "AAPL": "Apple Inc.",
    "MSFT": "Microsoft",
    "GOOGL": "Alphabet (Google)",
    "AMZN": "Amazon",
    "TSLA": "Tesla",
    "JPM": "JPMorgan Chase",
    "NKE": "Nike",
    "KO": "Coca-Cola",
    "VTI": "Vanguard Total Stock Market (ETF)"
}

# ============================================
# 2. Define the time period (3 years)
# ============================================
end_date = datetime.now()
start_date = end_date - timedelta(days=3*365) # 3 years of data

print(f"📅 Downloading data from {start_date.strftime('%Y-%m-%d')} to {end_date.strftime('%Y-%m-%d')}")

# ============================================
# 3. Download data for each company
# ============================================
all_data = []

for symbol, name in companies.items():
    print(f"⬇️ Downloading {name} ({symbol})...")
    
    # Download data
    ticker = yf.Ticker(symbol)
    df = ticker.history(start=start_date, end=end_date)
    
    # Reset index to make 'Date' a column
    df = df.reset_index()
    
    # Add company name and symbol
    df['Company'] = name
    df['Symbol'] = symbol
    
    # Select only the columns we need
    df = df[['Date', 'Symbol', 'Company', 'Open', 'High', 'Low', 'Close', 'Volume']]
    
    # Add to list
    all_data.append(df)
    
    print(f" ✅ Downloaded {len(df)} rows")

# ============================================
# 4. Combine all data into one DataFrame
# ============================================
final_df = pd.concat(all_data, ignore_index=True)

print(f"\n📊 Total data: {len(final_df)} rows")
print(f"📊 Date range: {final_df['Date'].min()} to {final_df['Date'].max()}")
print(f"📊 Companies: {final_df['Company'].nunique()}")

# ============================================
# 5. Save to CSV
# ============================================
final_df.to_csv('stock_data.csv', index=False)

print("\n✅ Data saved to 'stock_data.csv'")
print("\n📋 Sample data:")
print(final_df.head())
```
